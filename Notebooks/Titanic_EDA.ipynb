## Checking Missing Value

```python
print("Identify Column Missing value:")
df.isnull().any()                           ##  It's verify in column it's have be null value and get output as True and False.
Missing_Num_Value=df.isnull().sum()                           ##  It Count how many are Null value in a Column
Missing_Num_Value
```

```python
Missing_Value=df.isnull().sum()
Missing_Value

plt.figure(figsize=(11,9))
plt.xlabel("Column")
plt.ylabel("Count of Missing Value")
plt.title("Missing Value Count in Each Column")
plt.xticks(rotation=90)                                         ## Using rotate should not overlapping
sns.barplot(color="salmon",data=Missing_Value)


## Add Counts Numbers On Top of Each Bar

for bar in plt.gca().patches:
    # Get the height of the current bar
    height = bar.get_height()
    
    # Write the number on top of the bar
    plt.text(
        x = bar.get_x() + bar.get_width() / 2,        # Use bar.get_x() + bar.get_width()/2
        y = height,                                   # Use the height
        s = int(height),                              # Convert height into integer/string
        ha = 'center',                                # horizontal alignment
        va = 'bottom'                                  # vertical alignment
    )
```

## Data Cleaning

```python
# Check missing values in Age
print("Missing Age before:", df['Age'].isna().sum())

## Fill Missing age with a resonable value.

Median_age=df['Age'].median()
df['Age']=df['Age'].fillna(Median_age)

## Check Again
print("Missing age After: ",df['Age'].isna().sum())
```

```python
## Embarked Column have Categorical value like 'C','Q','S'.

## Check the Missing Value in Embarked
print("Missing Embarked Value :",df['Embarked'].isna().sum())

## Fill Missing Embarked Column 
Mode_Embarked=df['Embarked'].mode()[0]          ## 0 represent take a Single string like 'S','C','Q'
print("Mode of Embarked:",Mode_Embarked)

df['Embarked']=df['Embarked'].fillna(Mode_Embarked)

## Check Again:
print("Missing Embarked Value After:",df['Embarked'].isna().sum())
```

```python
## Check Missing Value in cabin

print("Missing cabin before:",df['Cabin'].isna().sum())
print("Total Rows:",len(df))
print("Fraction missing Cabin:",df['Cabin'].isna().mean())

## Drop The Cabin Table

df=df.drop(columns=['Cabin'])

# Confirm it's gone
print("Cabin in columns now:", 'Cabin' in df.columns)
```

```python
## Drop Unnecessary Ticket Column

## Check Current Columns:
print("Column before Dropping:",df.columns.to_list())

## Dropping Ticket Column:
df=df.drop(columns=['Ticket'])

## Check After Dropping column:
print("Column After Dropping:",df.columns.to_list())
```

## Univariate Analysis

```python
## Count of Survived and Not Survived by Countplot

df['Survived_label']=df['Survived'].map({0:"Survived",1:"Not Survived"})
custom_palette={'Survived':'crimson','Not Survived':'forestgreen'}

plt.figure(figsize=(10,7))
plt.title("Count of Survived v/s Not Survived")
plt.ylabel("Passenger Count")
plt.xlabel("Survived Status")
sns.countplot(x='Survived_label',hue='Survived_label',data=df,palette=custom_palette,legend=False)
```

```python
## Distibution of Fare

plt.title("Distibution of Fare")
sns.histplot(df['Fare'],bins=30)
```

```python
## Count of Passengers by Sex

plt.figure(figsize=(10,7))
plt.title("Count of Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Passanger Count")

custom_palette_1= {'male':'darkorange','female':'royalblue'}
sns.countplot(x='Sex',hue='Sex',data=df,palette=custom_palette_1,legend=False)
plt.show()
```
